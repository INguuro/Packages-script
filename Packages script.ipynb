```python
import pandas as pd

print(pd.__version__)


df = pd.read_csv(r"C:\Users\Admin\Desktop\Packages_Project.csv")

print(df.head())
```

```python
print(df.shape)          # Number of rows and columns
print(df.columns)        # Column names
print(df.info())         # Data types and non-null counts
print(df.isnull().sum()) # Missing values per column
```

```python
plt.title("Holiday Pakages")
```

```python
pivot_df = df.pivot_table(
    index="Category",
    columns="DurationDays",
    values="BasePriceUSD",
    aggfunc="mean",
    fill_value=0,
)
print(pivot_df)
```

```python
values_array = pivot_df.to_numpy()
reshaped_array = values_array.reshape(-1, 1)
```

```python
print("\n--- Pivot Table (Category vs Duration Days) ---")
print(pivot_df)
```

```python
df.to_csv("../pivot_table/processed_packages.csv", index=False)
pivot_df.to_csv("../pivot_table/category_duration_pivot.csv")
```

```python
df["BookingValueUSD"] = df["BasePriceUSD"]
df["ProfitUSD"] = df["BasePriceUSD"] - df["CostPerPaxUSD"]
df["ProfitMargin"] = df["ProfitUSD"] / df["BasePriceUSD"]

print(df["BookingValueUSD"])
```

```python
top_3_packages = df.nlargest(3, "ProfitUSD")
print("--- Top 3 Performing Packages ---")
print(
    top_3_packages[
        ["PackageID", "PackageName", "Category", "BasePriceUSD", "ProfitUSD"]
    ]
)
```

```python
plt.figure(figsize=(10,8))
```

```python
plt.title("Holiday Package Prics An &Days", fontsize=14, pad=15)
plt.xlabel("Duration(Days)", fontsize=12)
plt.ylabel("Package", fontsize=12)
```

```python
plt.tight_layout()
plt.show()
```

```python


# 1. Plot the pivot table as a bar chart
pivot_df.plot(kind="bar", figsize=(10, 6), edgecolor="black")

# 2. Add titles and labels
plt.title("Average Base Price by Category and Duration Days", fontsize=14, pad=15)
plt.xlabel("Category", fontsize=12)
plt.ylabel("Average Base Price (USD)", fontsize=12)
plt.xticks(rotation=0)
plt.legend(title="Duration Days", bbox_to_anchor=(1.05, 1), loc="upper left")

# 3. Show the plot
plt.tight_layout()
plt.show()
```
